# Code below will create missing derived HR dataset for years using EAP files, 2014 - 2024

```python
# 1. The exact 27 columns from your 2022 derived file
derived_2022_columns = [
    'UNITID', 'SALTOTL', 'SALPROF', 'SALASSC', 'SALASST', 'SALINST', 
    'SALLECT', 'SALNRNK', 'SFTETOTL', 'SFTEPSTC', 'SFTEINST', 'SFTERSRC', 
    'SFTEPBSV', 'SFTELCAI', 'SFTELCA', 'SFTEOTIS', 'SFTEMNGM', 'SFTEBFO', 
    'SFTECES', 'SFTECLAM', 'SFTEHLTH', 'SFTEOTHR', 'SFTESRVC', 'SFTESALE', 
    'SFTEOFAS', 'SFTENRCM', 'SFTEPTMM'
]

# 2. Map the EAPCAT numeric codes to the 2022 Column Names
category_mapping = {
    10000: 'SFTETOTL',
    21000: 'SFTEINST',
    22000: 'SFTERSRC',
    23000: 'SFTEPBSV',
    30000: 'SFTEMNGM',
    31000: 'SFTEBFO',
    32000: 'SFTECES',
    33000: 'SFTECLAM',
    34000: 'SFTEHLTH',
    35000: 'SFTESRVC',
    36000: 'SFTESALE',
    37000: 'SFTEOFAS',
    38000: 'SFTENRCM',
    39000: 'SFTEPTMM'
}

# 3. Loop through 2014 to 2020 (range is inclusive of start, exclusive of end)
years = range(2014, 2021)

for year in years:
    input_file = f'../Data/epa14-20/eap{year}_rv.csv'
    output_file = f'../Data/epa14-20/drvhr/drvhr{year}_rv.csv'
    
    # Safely skip years you haven't downloaded yet
    if not os.path.exists(input_file):
        print(f"[-] Missing: {input_file}. Please download it from IPEDS. Skipping...")
        continue
        
    print(f"[+] Processing {input_file}...")
    
    # Load raw data
    eap = pd.read_csv(input_file)
    
    # Filter for only the categories in our mapping dictionary
    target_data = eap[eap['EAPCAT'].isin(category_mapping.keys())].copy()
    
    # Clean the columns to ensure they are numeric
    target_data['EAPFT'] = pd.to_numeric(target_data['EAPFT'], errors='coerce').fillna(0)
    target_data['EAPPT'] = pd.to_numeric(target_data['EAPPT'], errors='coerce').fillna(0)
    
    # Calculate the Federal FTE Formula: Full-Time + (Part-Time / 3)
    target_data['FTE_CALC'] = np.round(target_data['EAPFT'] + (target_data['EAPPT'] / 3))
    
    # Pivot the table to get UNITID on the rows
    hr_simulated = target_data.pivot_table(
        index='UNITID',
        columns='EAPCAT',
        values='FTE_CALC',
        aggfunc='sum'
    ).reset_index()
    
    # Rename the columns using our dictionary
    hr_simulated = hr_simulated.rename(columns=category_mapping)
    
    # --- PAD THE MISSING COLUMNS ---
    # Add any missing columns (like the SALARY columns) and fill them with blanks/NaNs
    for col in derived_2022_columns:
        if col not in hr_simulated.columns:
            hr_simulated[col] = np.nan
            
    # Reorder columns to match the 2022 file exactly
    hr_simulated = hr_simulated[derived_2022_columns]
    
    # Save to CSV
    hr_simulated.to_csv(output_file, index=False)
    print(f"    -> Successfully saved {output_file} with {len(hr_simulated.columns)} columns!\n")
    
print("Data pipeline execution complete for 2014-2020!")
```

```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path('../Data/raw_data')
YEARS = range(2014, 2025)


def read_ipeds(path):
    """Read an IPEDS csv and normalize headers.

    Encoding varies across years: some files are UTF-8 (a few with a BOM), others are
    Windows-encoded with bytes cp1252 cannot decode. utf-8-sig handles the BOM case and
    latin-1 decodes any byte, so the pair covers every file without failing.
    """
    for encoding in ('utf-8-sig', 'latin-1'):
        try:
            df = pd.read_csv(path, low_memory=False, encoding=encoding)
            break
        except UnicodeDecodeError:
            continue
    df.columns = df.columns.str.strip()
    return df


def pick(folder, *names):
    """First existing file, preferring the _rv (revised) version."""
    for name in names:
        path = folder / name
        if path.exists():
            return path
    return None


for year in YEARS:
    year_dir = BASE_DIR / str(year)
    source_dir = year_dir / 'unzipped' if (year_dir / 'unzipped').exists() else year_dir
    if not source_dir.exists():
        print(f"Skipping {year}: folder not found -> {source_dir}")
        continue

    curr, prev = str(year)[-2:], str(year - 1)[-2:]
    master_file    = pick(source_dir, f'hd{year}.csv')
    financial_file = pick(source_dir, f'f{prev}{curr}_f1a_rv.csv', f'f{prev}{curr}_f1a.csv')
    hr_file        = pick(source_dir, f'drvhr{year}_rv.csv', f'drvhr{year}.csv')

    missing = [n for n, p in [('hd', master_file), ('f1a', financial_file), ('drvhr', hr_file)]
               if p is None]
    if missing:
        print(f"Skipping {year}: missing {missing} in {source_dir}")
        continue

    master_df = read_ipeds(master_file)
    public_institution_df = master_df.loc[master_df['SECTOR'] == 1].copy()

    join_df = (public_institution_df
               .merge(read_ipeds(financial_file), how='left', on='UNITID')
               .merge(read_ipeds(hr_file), how='left', on='UNITID'))

    # Write to the year folder (not the unzipped subfolder).
    output_file = year_dir / f'join_df_{year}.csv'
    join_df.to_csv(output_file, index=False)
    print(f"Saved {output_file}  shape={join_df.shape}")
```

# Add Fall Enrollment (EF) data to the yearly join_df files, 2014-2024

Enrollment is the driver our finance-only panel could not see. Two EF files are merged per year:

- **`ef{year}a`** - fall headcount. This file is **long** (one row per institution x `EFALEVEL`), so it is pivoted to one row per institution before merging.
- **`ef{year}d`** - retention rates and student-to-faculty ratio, already one row per institution.

Files are `_rv` (revised) through 2023 and un-revised for 2024, so the loader prefers `_rv` and falls back to the plain name. The cell is safe to re-run: existing EF columns are dropped before re-merging.

```python
EF_DIR = Path('../Data/ef')
BASE_DIR = Path('../Data/raw_data')
YEARS = range(2014, 2025)

# EF__A is long: one row per UNITID x EFALEVEL. Keep these level codes.
# (Verify the codes against the EF dictionary for your years before relying on them.)
EFALEVEL_MAP = {1: 'EF_TOTAL', 2: 'EF_UNDERGRAD', 12: 'EF_GRADUATE'}

# EF__D is one row per institution; keep these columns if present.
EFD_KEEP = ['RET_PCF', 'RET_PCP', 'STUFACR']


def join_path_for(year):
    """join_df is written to the year folder; also check unzipped/ for files from older runs."""
    for path in (BASE_DIR / str(year) / f'join_df_{year}.csv',
                 BASE_DIR / str(year) / 'unzipped' / f'join_df_{year}.csv'):
        if path.exists():
            return path
    return None


def ef_path(year, letter):
    """EF files are *_rv.csv through 2023 and un-revised for 2024."""
    for name in (f'ef{year}{letter}_rv.csv', f'ef{year}{letter}.csv'):
        path = EF_DIR / name
        if path.exists():
            return path
    return None


def load_ef_a(year):
    """Fall enrollment headcount, pivoted from long to one row per institution."""
    path = ef_path(year, 'a')
    if path is None:
        print(f"    [-] no ef{year}a file found")
        return None
    df = pd.read_csv(path, low_memory=False, encoding='cp1252')
    df.columns = df.columns.str.strip().str.upper()
    if 'EFALEVEL' not in df.columns or 'EFTOTLT' not in df.columns:
        print(f"    [!] {path.name}: EFALEVEL/EFTOTLT missing, skipping")
        return None
    subset = df[df['EFALEVEL'].isin(EFALEVEL_MAP)]
    wide = (subset.pivot_table(index='UNITID', columns='EFALEVEL',
                               values='EFTOTLT', aggfunc='sum')
                  .rename(columns=EFALEVEL_MAP)
                  .reset_index())
    wide.columns.name = None
    return wide


def load_ef_d(year):
    """Retention rates and student-to-faculty ratio (already one row per institution)."""
    path = ef_path(year, 'd')
    if path is None:
        print(f"    [-] no ef{year}d file found")
        return None
    df = pd.read_csv(path, low_memory=False, encoding='cp1252')
    df.columns = df.columns.str.strip().str.upper()
    missing = [c for c in EFD_KEEP if c not in df.columns]
    if missing:
        print(f"    [!] {path.name}: columns not found -> {missing}")
    keep = ['UNITID'] + [c for c in EFD_KEEP if c in df.columns]
    return df[keep].drop_duplicates('UNITID')


EF_COLS = list(EFALEVEL_MAP.values()) + EFD_KEEP

for year in YEARS:
    join_path = join_path_for(year)
    if join_path is None:
        print(f"Skipping {year}: no join_df_{year}.csv found in "
              f"{BASE_DIR / str(year)} or its unzipped/ subfolder")
        continue

    join_df = pd.read_csv(join_path, low_memory=False)
    # Drop any EF columns from a previous run so this cell is safe to re-execute.
    join_df = join_df.drop(columns=[c for c in EF_COLS if c in join_df.columns])
    start_cols, start_rows = join_df.shape[1], join_df.shape[0]

    for loader in (load_ef_a, load_ef_d):
        block = loader(year)
        if block is not None:
            join_df = join_df.merge(block, how='left', on='UNITID')

    # A merge that changes the row count means duplicate UNITIDs slipped through.
    assert join_df.shape[0] == start_rows, f"{year}: row count changed on EF merge"

    join_df.to_csv(join_path, index=False)
    coverage = {c: int(join_df[c].notna().sum()) for c in EF_COLS if c in join_df.columns}
    print(f"{year}: {start_cols} -> {join_df.shape[1]} cols | rows {join_df.shape[0]} | "
          f"non-null {coverage} | {join_path.parent}")

print("\nEF merge complete.")
```

# Add EAP faculty-tenure detail to the yearly join_df files

IPEDS EAP (Employees by Assigned Position) gives instructional headcount by tenure status, which lets us measure **adjunctification** - the shift toward contingent faculty that the Introduction names as a harm (Ehrenberg & Zhang 2005; Bettinger & Long 2010; AAUP 2025). Instructional staff are `OCCUPCAT == 210`; tenure status is `FACSTAT` (20 tenured, 30 tenure-track, 40 not-on-track).

**Coverage caveat:** the available files span **2014-2020 only**. Rows for 2021-2024 will have blank tenure columns, which matters because the rolling-origin test years (2021, 2022) fall in that gap - so these features cannot influence the held-out evaluation until the 2021-2024 EAP files are added.

```python
EAP_DIR = Path('../Data/epa14-20')

# Instructional staff (OCCUPCAT 210) by tenure status (FACSTAT):
#   0 = total, 20 = tenured, 30 = on tenure track, 40 = not on track.
FACSTAT_MAP = {20: 'EAP_INST_TENURED', 30: 'EAP_INST_TENURETRACK', 40: 'EAP_INST_NOTRACK'}
EAP_COLS = ['EAP_INST_TOTAL', 'EAP_INST_PT'] + list(FACSTAT_MAP.values())


def eap_path(year):
    for name in (f'eap{year}_rv.csv', f'eap{year}.csv'):
        path = EAP_DIR / name
        if path.exists():
            return path
    return None


def load_eap(year):
    path = eap_path(year)
    if path is None:
        print(f"    [-] no EAP file for {year}")
        return None
    df = read_ipeds(path)
    df.columns = df.columns.str.upper()
    inst = df[df['OCCUPCAT'] == 210]
    if inst.empty:
        print(f"    [!] {path.name}: no OCCUPCAT==210 rows")
        return None
    total = (inst[inst['FACSTAT'] == 0]
             .groupby('UNITID')
             .agg(EAP_INST_TOTAL=('EAPTOT', 'sum'), EAP_INST_PT=('EAPPT', 'sum'))
             .reset_index())
    status = (inst[inst['FACSTAT'].isin(FACSTAT_MAP)]
              .pivot_table(index='UNITID', columns='FACSTAT', values='EAPTOT', aggfunc='sum')
              .rename(columns=FACSTAT_MAP).reset_index())
    status.columns.name = None
    return total.merge(status, how='outer', on='UNITID')


for year in YEARS:
    join_path = join_path_for(year)
    if join_path is None:
        print(f"Skipping {year}: no join_df file found")
        continue

    join_df = pd.read_csv(join_path, low_memory=False)
    join_df = join_df.drop(columns=[c for c in EAP_COLS if c in join_df.columns])
    start_cols, start_rows = join_df.shape[1], join_df.shape[0]

    block = load_eap(year)
    if block is not None:
        join_df = join_df.merge(block, how='left', on='UNITID')

    assert join_df.shape[0] == start_rows, f"{year}: row count changed on EAP merge"
    join_df.to_csv(join_path, index=False)
    coverage = {c: int(join_df[c].notna().sum()) for c in EAP_COLS if c in join_df.columns}
    print(f"{year}: {start_cols} -> {join_df.shape[1]} cols | non-null {coverage or 'EAP absent'}")

print("\nEAP merge complete.")
```

# Combine the yearly files into one master panel, 2014-2024

Stacks every `join_df_{year}.csv` into a single institution-by-year panel tagged with `YEAR`. Run this after any new data source is merged above, so the master panel always reflects the latest additions.

Two things the naive `pd.concat` misses, handled here:
- **Header drift** - a few yearly files carry trailing spaces in column names (e.g. `SFTEPTMM   `), which would otherwise become separate columns. Names are stripped and duplicates collapsed.
- **Schema drift** - the yearly files do not share an identical column set, so `concat` takes the union and fills `NaN` where a year lacked a column. The cell reports how many columns are affected so this is visible rather than silent.

```python
frames = []
for year in YEARS:
    path = join_path_for(year)
    if path is None:
        print(f"[-] {year}: no join_df file found, skipping")
        continue

    df = pd.read_csv(path, low_memory=False)
    # Normalize headers so trailing-space variants collapse into one column.
    df.columns = df.columns.str.strip()
    df = df.loc[:, ~df.columns.duplicated()]
    frames.append(df.assign(YEAR=year))
    print(f"[+] {year}: {len(df)} rows, {df.shape[1]} cols")

assert frames, "No yearly files loaded -- check BASE_DIR."
master_panel_df = pd.concat(frames, ignore_index=True)

# One row per institution-year.
before = len(master_panel_df)
master_panel_df = (master_panel_df
                   .drop_duplicates(['UNITID', 'YEAR'])
                   .sort_values(['UNITID', 'YEAR'])
                   .reset_index(drop=True))
if len(master_panel_df) < before:
    print(f"\ndropped {before - len(master_panel_df)} duplicate (UNITID, YEAR) rows")

# How much of the width is an artifact of schema drift across years?
column_sets = [set(f.columns) for f in frames]
in_all_years = set.intersection(*column_sets)
print(f"\nMaster panel: {master_panel_df.shape[0]} institution-years x {master_panel_df.shape[1]} cols")
print(f"  years        : {master_panel_df['YEAR'].min()}-{master_panel_df['YEAR'].max()} "
      f"({master_panel_df['YEAR'].nunique()} loaded)")
print(f"  institutions : {master_panel_df['UNITID'].nunique()}")
print(f"  columns present in every year: {len(in_all_years)} "
      f"({master_panel_df.shape[1] - len(in_all_years)} appear in only some years)")
```

## Build the prediction target

`TARGET_STAFF_CUT` = a decline of 5% or more in instructional FTE staff over a **two-year** horizon (`t -> t+2`), which matches the 18-24 month lead time the project promises. The label is forward-looking, computed per institution across a gap-free window, and left as `<NA>` where no year `t+2` exists - those rows are unlabeled, not negative.

`TARGET_VULNERABLE_TOTAL` (any decline in total staff) is kept as a secondary label for comparison.

```python
STAFF_COL = 'SFTEINST'      # instructional FTE staff
MIN_BASE  = 1.0             # floor on the base year so tiny denominators don't distort the ratio
HORIZON   = 2               # years ahead: 2 = a cut over t -> t+2

group = master_panel_df.groupby('UNITID')


def forward_pct(col, horizon=HORIZON):
    """Forward % change (t -> t+horizon), valid only across a gap-free window with a usable base."""
    current = master_panel_df[col]
    future = group[col].shift(-horizon)
    future_year = group['YEAR'].shift(-horizon)
    valid = ((future_year - master_panel_df['YEAR']).eq(horizon)
             & current.notna() & future.notna() & current.ge(MIN_BASE))
    return ((future - current) / current * 100).where(valid), valid


pct_inst, valid_inst = forward_pct(STAFF_COL)
pct_totl, valid_totl = forward_pct('SFTETOTL')

master_panel_df = master_panel_df.assign(
    FTE_INST_PCT_CHANGE_2Y  = pct_inst,
    FTE_TOTL_PCT_CHANGE_2Y  = pct_totl,
    TARGET_STAFF_CUT        = (pct_inst <= -5).where(valid_inst).astype('Int64'),
    TARGET_VULNERABLE_TOTAL = (pct_totl < 0).where(valid_totl).astype('Int64'),
)

for target in ['TARGET_STAFF_CUT', 'TARGET_VULNERABLE_TOTAL']:
    labeled = master_panel_df[target].notna()
    print(f"{target}: labeled={int(labeled.sum())} of {len(master_panel_df)}, "
          f"positives={int(master_panel_df[target].sum())}, "
          f"rate={master_panel_df.loc[labeled, target].mean() * 100:.1f}%")

labeled_years = master_panel_df.loc[master_panel_df['TARGET_STAFF_CUT'].notna(), 'YEAR']
print(f"\nlabeled predictor years: {labeled_years.min()}-{labeled_years.max()} "
      f"(the last {HORIZON} years cannot be labeled yet)")

master_panel_df[['UNITID', 'YEAR', STAFF_COL, 'FTE_INST_PCT_CHANGE_2Y', 'TARGET_STAFF_CUT']].head(10)
```
