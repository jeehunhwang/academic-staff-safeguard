# Code below will create missing derived HR dataset for years using EAP files, 2014 - 2024

```python
import pandas as pd
import numpy as np
import os

# 1. The exact 27 columns from your 2022 derived file
derived_2022_columns = [
    'UNITID', 'SALTOTL', 'SALPROF', 'SALASSC', 'SALASST', 'SALINST', 
    'SALLECT', 'SALNRNK', 'SFTETOTL', 'SFTEPSTC', 'SFTEINST', 'SFTERSRC', 
    'SFTEPBSV', 'SFTELCAI', 'SFTELCA', 'SFTEOTIS', 'SFTEMNGM', 'SFTEBFO', 
    'SFTECES', 'SFTECLAM', 'SFTEHLTH', 'SFTEOTHR', 'SFTESRVC', 'SFTESALE', 
    'SFTEOFAS', 'SFTENRCM', 'SFTEPTMM'
]

# 2. Map the EAPCAT numeric codes to the 2022 Column Names
category_mapping = {
    10000: 'SFTETOTL',
    21000: 'SFTEINST',
    22000: 'SFTERSRC',
    23000: 'SFTEPBSV',
    30000: 'SFTEMNGM',
    31000: 'SFTEBFO',
    32000: 'SFTECES',
    33000: 'SFTECLAM',
    34000: 'SFTEHLTH',
    35000: 'SFTESRVC',
    36000: 'SFTESALE',
    37000: 'SFTEOFAS',
    38000: 'SFTENRCM',
    39000: 'SFTEPTMM'
}

# 3. Loop through 2014 to 2020 (range is inclusive of start, exclusive of end)
years = range(2014, 2021)

for year in years:
    input_file = f'../Data/epa14-20/eap{year}_rv.csv'
    output_file = f'../Data/epa14-20/drvhr/drvhr{year}_rv.csv'
    
    # Safely skip years you haven't downloaded yet
    if not os.path.exists(input_file):
        print(f"[-] Missing: {input_file}. Please download it from IPEDS. Skipping...")
        continue
        
    print(f"[+] Processing {input_file}...")
    
    # Load raw data
    eap = pd.read_csv(input_file)
    
    # Filter for only the categories in our mapping dictionary
    target_data = eap[eap['EAPCAT'].isin(category_mapping.keys())].copy()
    
    # Clean the columns to ensure they are numeric
    target_data['EAPFT'] = pd.to_numeric(target_data['EAPFT'], errors='coerce').fillna(0)
    target_data['EAPPT'] = pd.to_numeric(target_data['EAPPT'], errors='coerce').fillna(0)
    
    # Calculate the Federal FTE Formula: Full-Time + (Part-Time / 3)
    target_data['FTE_CALC'] = np.round(target_data['EAPFT'] + (target_data['EAPPT'] / 3))
    
    # Pivot the table to get UNITID on the rows
    hr_simulated = target_data.pivot_table(
        index='UNITID',
        columns='EAPCAT',
        values='FTE_CALC',
        aggfunc='sum'
    ).reset_index()
    
    # Rename the columns using our dictionary
    hr_simulated = hr_simulated.rename(columns=category_mapping)
    
    # --- PAD THE MISSING COLUMNS ---
    # Add any missing columns (like the SALARY columns) and fill them with blanks/NaNs
    for col in derived_2022_columns:
        if col not in hr_simulated.columns:
            hr_simulated[col] = np.nan
            
    # Reorder columns to match the 2022 file exactly
    hr_simulated = hr_simulated[derived_2022_columns]
    
    # Save to CSV
    hr_simulated.to_csv(output_file, index=False)
    print(f"    -> Successfully saved {output_file} with {len(hr_simulated.columns)} columns!\n")
    
print("Data pipeline execution complete for 2014-2020!")
```

```python
import pandas as pd
from pathlib import Path

# Base directory containing year folders (e.g., ../Data/raw_data/2024/unzipped)
base_dir = Path('../Data/raw_data')

# Loop from 2024 down to 2014
for year in range(2014, 2021):
    unzipped_dir = base_dir / str(year) / 'unzipped'

    if not unzipped_dir.exists():
        print(f"Skipping {year}: folder not found -> {unzipped_dir}")
        continue

    # Assumes files are named like:
    #   hd2024.csv
    #   f2224_f1a.csv
    #   drvhr2024.csv
    curr_year = str(year)[-2:]  # This line seems to be incomplete or a typo. It should probably be removed or corrected.
    prev_year = str(year - 1)[-2:]  # This line seems to be incomplete or a typo. It should probably be removed or corrected.

    master_file = unzipped_dir / f'hd{year}.csv'
    financial_file = unzipped_dir / f'f{prev_year}{curr_year}_f1a_rv.csv'
    hr_file = unzipped_dir / f'drvhr{year}_rv.csv'

    if not master_file.exists():
        print(f"Skipping {year}: missing {master_file}")
        continue
    if not financial_file.exists():
        print(f"Skipping {year}: missing {financial_file}")
        continue
    if not hr_file.exists():
        print(f"Skipping {year}: missing {hr_file}")
        continue

    # Read and prepare data
    master_df = pd.read_csv(master_file, low_memory=False, encoding='cp1252')
    public_institution_df = master_df.loc[master_df['SECTOR'] == 1].copy()

    financial_df = pd.read_csv(financial_file, low_memory=False, encoding='cp1252')
    hr_df = pd.read_csv(hr_file, low_memory=False, encoding='cp1252')

    # Merge dataframes
    helper_df = public_institution_df.merge(financial_df, how='left', on='UNITID')
    join_df = helper_df.merge(hr_df, how='left', on='UNITID')

    # Save join_df into the corresponding year's unzipped folder
    output_file = unzipped_dir / f'join_df_{year}.csv'
    join_df.to_csv(output_file, index=False)

    print(f"Saved {output_file} with shape {join_df.shape}")
```
